```python
def summary(player_id : int):
    dfp = df[(df['batter']==player_id) & (df['out'] != 'DIDNOTBAT')]
    innings : int = len(dfp)
    runs : int = dfp['runs'].sum()
    times_out = len(dfp[(dfp['out'] != 'DIDNOTBAT') & (dfp['out'] != 'NOTOUT')])
    not_out = len(dfp[dfp['out'] == 'NOTOUT'])
    avg = round(runs/times_out,2)
    hundreds = len(dfp[df['runs'] > 99])
    fifties = len((dfp[(df['runs'] > 49) & (df['runs'] < 100)]))
    highest_score = dfp['runs'].max()
    fours = dfp['fours'].sum()
    sixes = dfp['sixes'].sum()
    return innings,not_out,runs,highest_score,avg,hundreds,fifties,fours,sixes
    
    
```

```python
f = open('logs/player.stats.csv','w')
f.write("name,innings,not_out,runs,highest_score,avg,hundreds,fifties,fours,sixes\n")
d = {25:'Sachin Tendulkar',1413:'Virat Kohli',3817:'Sunil Gavaskar',3832:'Dilip Vengsarkar',4160:'Sir Garry Sobers',4163:'Sir Viv Richards',3664:'Javed Miandad',5321:'Sir Ian Botham',3647:'Imran Khan',8019:'Joe Root',2250:'Steven Smith',6326:'Kane Williamson'}
for id,name in d.items():
    innings,not_out,runs,highest_score,avg,hundreds,fifties,fours,sixes = summary(id)
    f.write(f"{name},{innings},{not_out},{runs},{highest_score},{avg},{hundreds},{fifties},{fours},{sixes}\n")

f.close()
print('DONE')
```
