```python
import pandas as pd
from pandas import DataFrame
import numpy as np
from pandas.api.types import is_numeric_dtype
```

```python
df = pd.read_csv('data/batting.csv')
df.info()
```

```python
def optimize_df(do_optimize = False):
    for col in df.columns:
        if is_numeric_dtype(df[col]):
            if max(df[col]) < np.iinfo(np.uint8).max:
                if do_optimize:
                    df[col] = df[col].astype(np.uint8)
                    print(f"{col} is optimized to: {np.uint8}")
                else:
                    print(f"{col} can be optimized to: {np.uint8}")
            else:
                if np.iinfo(np.uint8).max < max(df[col]) < np.iinfo(np.uint16).max:
                    if do_optimize:
                        df[col] = df[col].astype(np.uint16)
                        print(f"{col} is optimized to: {np.uint16}")
                    else:
                        print(f"{col} can be optimized to: {np.uint16}")
                else:
                    if np.iinfo(np.uint16).max < max(df[col]) < np.iinfo(np.uint32).max:
                        if do_optimize:
                            print(f"{col} is optimized to: {np.uint32}")
                            df[col] = df[col].astype(np.uint32)
                        else:
                            print(f"{col} can be optimized to: {np.uint16}")
                    else:
                        raise Exception("Missed everything")
```

```python
optimize_df(True)
df.info()
```
